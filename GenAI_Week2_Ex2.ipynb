Different types of regularization techniques:
1. L1 Regularization
2. L2 Regularization using dropouts
3. Batch Normalization technique
4. Early Droping technique

Import Libraries

```python
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf

from tensorflow.keras.datasets import mnist
from tensorflow.keras.layers import *
from tensorflow.keras.models import Model
from tensorflow.keras import regularizers
```

Load Dataset

```python
(x_train, _), (x_test, _) = mnist.load_data()

x_train = x_train.astype("float32")/255.
x_test = x_test.astype("float32")/255.

x_train = x_train.reshape(-1,784)
x_test = x_test.reshape(-1,784)
```

L1 Regularization

```python
inp=Input(shape=(784,))

x=Dense(
    128,
    activation='relu',
    activity_regularizer=regularizers.l1(1e-4)
)(inp)

out=Dense(784, activation='sigmoid')(x)

model_l1=Model(inp,out)

model_l1.compile(
    optimizer='adam',
    loss='binary_crossentropy'
)
```

```python
model_l1.fit(
    x_train,
    x_train,
    epochs=5,
    batch_size=256,
    validation_data=(x_test,x_test)
)
```

2. L2 Regularization

```python
inp=Input(shape=(784,))

x=Dense(
    128,
    activation='relu',
    activity_regularizer=regularizers.l2(1e-4)
)(inp)

out=Dense(784, activation='sigmoid')(x)

model_l2=Model(inp,out)

model_l2.compile(
    optimizer='adam',
    loss='binary_crossentropy'
)
```

```python
model_l2.fit(
    x_train,
    x_train,
    epochs=5,
    batch_size=256,
    validation_data=(x_test,x_test)
)
```

3. Dropout Regularization

```python
inp=Input(shape=(784,))

x=Dense(256,activation='relu')(inp)
x=Dropout(0.3)(x)

x=Dense(128,activation='relu')(x)
x=Dropout(0.3)(x)

out= Dense(784,activation='relu')(x)

model_dropout=Model(inp,out)

model_dropout.compile(
    optimizer='adam',
    loss='binary_crossentropy'
)
```

```python
model_dropout.fit(
    x_train,
    x_train,
    epochs=5,
    batch_size=256,
    validation_data=(x_test,x_test)
)
```

4. Batch Normalization

```python
inp=Input(shape=(784,))

x=Dense(256)(inp)
x=BatchNormalization()(x)
x=Activation('relu')(x)

x=Dense(128)(x)
x=BatchNormalization()(x)
x=Activation('relu')(x)

out= Dense(784,activation='sigmoid')(x)

model_bn=Model(inp,out)

model_bn.compile(
    optimizer='adam',
    loss='binary_crossentropy'
)
```

```python
model_bn.fit(
    x_train,
    x_train,
    epochs=5,
    batch_size=256,
    validation_data=(x_test,x_test)
)
```

5. Early Stopping

```python
early_stop=tf.keras.callbacks.EarlyStopping(
    monitor='val_loss',
    patience=2,
    restore_best_weights=True
)
```

```python
inp=Input(shape=(784,))

x=Dense(256,activation='relu')(inp)
x=Dense(128,activation='sigmoid')(x)

out=Dense(784, activation='sigmoid')(x)

model_es=Model(inp,out)

model_es.compile(
    optimizer='adam',
    loss='binary_crossentropy'
)

model_es.fit(
    x_train,
    x_train,
    epochs=5,
    batch_size=256,
    validation_data=(x_test,x_test),
    callbacks=[early_stop]
)
```
